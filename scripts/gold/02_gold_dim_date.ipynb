# The Transformation Logic

```python

query = """
SELECT
  date,
  year(date)                              AS year,
  quarter(date)                           AS quarter,
  month(date)                             AS month,
  date_format(date, 'yyyy-MM')            AS year_month,
  concat(year(date), '-Q', quarter(date)) AS year_quarter
FROM (
  SELECT explode(sequence(DATE'2020-01-01', DATE'2030-12-31', INTERVAL 1 DAY)) AS date
);
"""

df = spark.sql(query)
```

# Write into Gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("humanoid_robotics.gold.dim_date")
     
```
