# The transformation logic

```python
query = """
SELECT
  event_id,
  doc_id,
  max(extracted_event_id = event_id) AS is_primary_source
FROM humanoid_robotics.silver.event_dedup_map
GROUP BY event_id, doc_id;
"""

df = spark.sql(query)
```

# Write in gold

```python

df.write.mode("overwrite").format("delta").saveAsTable("humanoid_robotics.gold.bridge_event_source")
     
```
