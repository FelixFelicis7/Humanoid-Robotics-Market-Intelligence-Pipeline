# The Transformation Logic

```python

query = """
SELECT
  d.doc_id,
  d.title,
  d.publisher,
  d.source_type,          -- press_release / filing / news / policy
  d.language,             -- en / zh
  d.publish_date,
  d.url,
  d.primary_company,      
  s.summary_en,
  coalesce(s.event_count, 0) AS event_count
FROM humanoid_robotics.silver.documents d
LEFT JOIN humanoid_robotics.silver.document_summaries s
  ON s.doc_id = d.doc_id
"""

df = spark.sql(query)
```

# Write into Gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("humanoid_robotics.gold.dim_source")
     
```
