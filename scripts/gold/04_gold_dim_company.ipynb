# The transformation Logic

```python
query = """
SELECT
  company_id,
  company_name,          
  company_name_zh,
  company_type,          -- humanoid_maker / tier1 / investor / customer / technology_partner / other     
  country,
  region
FROM humanoid_robotics.silver.company;
"""

df = spark.sql(query)
```

# Write into Gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("humanoid_robotics.gold.dim_company")
     
```
