# Initialization

```python
from pyspark.sql import functions as F
```

# Path Configuration

```python
CATALOG = "humanoid_robotics"
MANIFEST_PATH = f"/Volumes/{CATALOG}/landing/manifest/manifest_pilot.csv"
TARGET_TABLE  = f"{CATALOG}.bronze.manifest_pilot"
```

# Read manifest

```python
df_manifest = (
    spark.read
    .option("header", True)
    .option("multiLine", True)
    .option("escape", '"')
    .csv(MANIFEST_PATH)
)

df_manifest = df_manifest.toDF(*[c.replace("\ufeff", "").strip() for c in df_manifest.columns])

df_manifest.display()
```

# Ingest Files into bronze table

```python
(
        df_manifest.write
          .mode("overwrite")
          .format("delta")
          .option("overwriteSchema", True)
          .saveAsTable(TARGET_TABLE)
)
```
